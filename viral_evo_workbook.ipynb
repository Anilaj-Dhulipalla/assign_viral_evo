````python
#PROBLEM 1 CELL

import matplotlib.pyplot as plt
import re
from helper_functions import jukes_cantor


# Read FASTA file
records = list(SeqIO.parse("data/p1-sars-cov-2-variants.fasta", "fasta"))

data = []
for record in records: 
    desc = record.description
    parts = desc.split("|")
    accession = parts[0]
    country = parts[1] if len(parts) > 1 else "Unknown"

    match = re.search(r'\d{4}-\d{2}(-\d{2})?', desc)
    date_str = match.group(0) if match else None

    # Extract date
    data.append({
        "id": accession, 
        "country": country,
        "collection_date": date_str,
        "sequence": str(record.seq),
        "description": desc
    })

df = pd.DataFrame(data)

df['date'] = pd.to_datetime(df['collection_date'], format='mixed', errors='coerce')

df = df.dropna(subset=['date']).reset_index(drop=True)


# Retrieve reference sequence from NCBI 
handle = Entrez.efetch(db="nuccore", id="NC_045512.2", rettype="fasta", retmode="text")
ref_record = SeqIO.read(handle, "fasta")
ref_seq = str(ref_record.seq)
handle.close()

# identify earliest collected seqence as starting point
earliest_row = df.loc[df['date'].idxmin()]
start_date = earliest_row['date']
df['days'] = (df['date'] - start_date).dt.days

# calculate JC corrected genetic distance from reference/earliest sequence
distances = []
for seq in df['sequence']:
    d = jukes_cantor(ref_seq, seq)
    distances.append(d)
df['jc_distance'] = distances

# sort by days for regression
df = df.sort_values('days').reset_index(drop=True)

# linear regression using np.polyfit
slope, intercept = np.polyfit(df['days'], df['jc_distance'], 1)

sars_cov_2_per_genome_per_day = float(slope)
genome_length = len(ref_seq.replace("-", ""))
sars_cov_2_per_nt_per_day = sars_cov_2_per_genome_per_day / genome_length

print(f"Mutation rate per genome per day: {sars_cov_2_per_genome_per_day: .5f}")
print(f"Mutation rate per nucelotide per day: {sars_cov_2_per_nt_per_day: .7f}")

# plotting and saving to problem1.svg
plt.figure(figsize=(8,6))
plt.scatter(df['days'], df['jc_distance'], alpha=0.6, label='Viral Genomes')
plt.plot(df['days'], slope * df['days'] + intercept, color='red', label=f'Linear Fit (Rate: {sars_cov_2_per_genome_per_day:.4f}/day)')
plt.xlabel('Time Elapsed (Days)')
plt.ylabel('Jukes-Cantor Corrected Genetic Distance')
plt.title('SARS-CoV-2 Mutation Rate Over Time')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.savefig('problem1.svg', format='svg', bbox_inches='tight')
plt.show()
````

````python
need_for_JC_correction = """
Jukes-Cantor correction is needed as raw Hamming distances underestimates the true number of evolutionary changes as genetic divergence increases over time.
As viruses mutate, multiple substitutions can occur at the same nucleotide side (e.g. A -> G -> A), which raw distance metrics fail to capture. 
The JC correction applies a probabilistic adjustement to account for these hidden mutations, linearising the relationship between observed genetic differences and true evolutionary time.
"""
````

## Problem 3: Reinterpreting mutation rates

Majority of RNA viral genomes are single-stranded RNA, where only one strand of the viruses encode for the genes. We divide these viruses into positive-sense and negative-sense viruses, based on where they code for genes.
Positive-sense RNA directly codes for genes and acts like mRNA, meaning it can be directly used by the host's ribosomes to make proteins.
Negative-sense RNA is a reverse complement of the positive-sense RNA and does not code for genes directly. It must first be converted into positive-sense RNA by the virus's own enzymes before it can be used to make proteins.
Positive-sense RNA viruses mutate faster because their replication process is simpler and mostly lacks error correction. Negative-sense RNA viruses, on the other hand, mutate more slowly since they have an extra step in replication and require more functional genes to support replication.

## Problem 4: Variant-specific mutations

As the virus mutates, it inevitably evolves and proliferates around the world. Every so often, some mutations may prove especially beneficial to the spread of the virus, and this version of the virus spreads faster than other versions. When a version of a virus becomes especially prevalent inside a population, we call this a virus *variant*. Variants are nothing more than a naming scheme for viruses that have specific mutations. For instance, in Slovenia during 2022, we were dealing with the Omicron variant. Think of this as observing natural selection in real time. Some viruses have mutations that enable them to spread more easily throughout our population, which inevitably leads to the demise of other virus variants, which are not as good at proliferation. The result is survival of the fittest at the viral level, where, unfortunately, the fittest viruses seem to cause the most damage to us humans.

How do we identify variants? A variant is determined by several so-called *defining mutations*. Mutations can either be synonymous or nonsynonymous. Synonymous mutations are changes in nucleotide bases that result in the same encoded amino acid and are thus less important. Nonsynonymous mutations are nucleotide mutations that alter the amino acid sequence of a protein.

## Problem 5: Identifying variants

As we've seen, mutations are not distributed uniformly across the genome. Important mutations at specific sites distinguish variants from each other. Nonsynonymous mutations are the ones reflected in the proteins they produce.

A standard convention is to denote mutations using a short string, e.g., *S:T 19 R*. The first part denotes the protein where the mutation has occurred. The second part is comprised of the actual mutation. For instance, *S:T 19 R* means that we are looking at a mutation on the S -- spike protein, where the original amino acid at location 19 was T -- threonine, which changed to R -- arginine. The "-" symbol indicates a deletion (e.g., *S:H 69 -*). For a complete reference of SARS-CoV-2 mutations, you can take a look at https://covariants.org and look through the different variants. We will not consider insertions in this homework, as it would only complicate our lives. Note that location of aminoacids in proteins start with 1 and not 0 like in python notation.

````python
# PROBlEM 5 CELL

import json

# Load variants.json
with open("data/variants.json", "r") as f:
    variants_data = json.load(f)

# parse variant nonsynonymous spike mutations
variant_mutations = {}
for v_name, v_info in variants_data.items():
    muts = {}
    for m in v_info.get("nonsynonymous", []):
        if m.get("gene") == "S":
            # 1-indexed position in json to 0-indexed 
            pos = int(m["pos"]) - 1
            expected_aa = m["right"]
            muts[pos] = expected_aa
    variant_mutations[v_name] = muts

# Parse spike protein FASTA
records = list(SeqIO.parse("data/p5-aus-spike-proteins.fasta", "fasta"))

record_metadata = []
for record in records:
    desc = record.description
    seq_str = str(record.seq).upper()
    
    # Extract GISAID accession ID 
    parts = desc.split('|')
    accession = parts[2].strip() if len(parts) > 2 else parts[0].strip()
    
    # Extract collection date (YYYY-MM-DD)
    match = re.search(r'(\d{4}-\d{2}-\d{2})', desc)
    date_str = match.group(1) if match else None
    
    record_metadata.append({
        "id": accession,
        "date_str": date_str,
        "sequence": seq_str
    })

df_seqs = pd.DataFrame(record_metadata)
df_seqs['date'] = pd.to_datetime(df_seqs['date_str'], errors='coerce')

# Establish baseline reference sequence 
df_seqs_sorted = df_seqs.sort_values('date').reset_index(drop=True)
base_ref_seq = df_seqs_sorted.loc[0, 'sequence'] if not df_seqs_sorted.empty else df_seqs.loc[0, 'sequence']

classification_rows = []

for idx, row in df_seqs.iterrows():
    seq = row['sequence']
    accession = row['id']
    
    # Check NO_MUTATIONS 
    has_any_mutation = False
    for i in range(min(len(seq), len(base_ref_seq))):
        char = seq[i]
        ref_char = base_ref_seq[i]
        if char == 'X' or ref_char == 'X':
            continue
        if char != ref_char:
            has_any_mutation = True
            break
            
    if not has_any_mutation:
        classification_rows.append((accession, "NO_MUTATIONS"))
        continue
        
    # Evaluate percentage match for each variant
    best_variant = "UNKNOWN"
    max_match_pct = 0.0
    
    for v_name, muts in variant_mutations.items():
        if len(muts) == 0:
            continue
        matched_count = 0
        total_variant_muts = len(muts)
        
        for pos, expected_aa in muts.items():
            if pos < len(seq):
                actual_aa = seq[pos]
                if actual_aa != 'X' and actual_aa == expected_aa:
                    matched_count += 1
                    
        match_pct = matched_count / total_variant_muts
        if match_pct > max_match_pct:
            max_match_pct = match_pct
            best_variant = v_name
            
    # Apply 51% threshold rule
    if max_match_pct >= 0.51:
        classification_rows.append((accession, best_variant))
    else:
        classification_rows.append((accession, "UNKNOWN"))

# Save classification CSV (Accession ID, Variant Name, no header)
df_class = pd.DataFrame(classification_rows)
df_class.to_csv("problem5-classification.csv", index=False, header=False)

# Create monthly distribution stackplot and save to problem5.svg
df_merged = pd.merge(df_seqs, df_class, left_on='id', right_on=0)
df_merged = df_merged.rename(columns={1: 'variant'})

df_merged = df_merged.dropna(subset=['date'])
df_merged['year_month'] = df_merged['date'].dt.to_period('M')

# Pivot counts and forward-fill to maintain a continuous monthly timeline without gaps
pivot_counts = df_merged.pivot_table(index='year_month', columns='variant', aggfunc='size', fill_value=0)
full_range = pd.period_range(start=pivot_counts.index.min(), end=pivot_counts.index.max(), freq='M')
pivot_counts = pivot_counts.reindex(full_range).ffill().fillna(0)

pivot_pcts = pivot_counts.div(pivot_counts.sum(axis=1), axis=0) * 100

# Plotting
fig, ax = plt.subplots(figsize=(14, 7))

months = [str(p) for p in pivot_pcts.index]
y_data = [pivot_pcts[col].values for col in pivot_pcts.columns]

ax.stackplot(months, y_data, labels=pivot_pcts.columns, alpha=0.85)

ax.set_title("SARS-CoV-2 Variant Distribution Over Time in Australia", fontsize=14, fontweight='bold')
ax.set_xlabel("Time (Month)", fontsize=12)
ax.set_ylabel("Percentage (%)", fontsize=12)
ax.set_ylim(0, 100)

# Clean up x-axis ticks
step = max(1, len(months) // 12)
ax.set_xticks(range(0, len(months), step))
ax.set_xticklabels([months[i] for i in range(0, len(months), step)], rotation=45, ha='right')

ax.legend(loc='center left', bbox_to_anchor=(1, 0.5), framealpha=0.9)
plt.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.savefig('problem5.svg', format='svg', bbox_inches='tight')
plt.show()
````
